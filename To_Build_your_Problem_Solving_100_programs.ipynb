```python
def check_palindrome_number(n: int) -> bool:
    temp = n
    rev_num = 0
    while temp > 0:
        digit = temp % 10
        rev_num = rev_num * 10 + digit
        temp //= 10
    return True if rev_num == n else False

num = 1234321
print(f"Is {num} is a Palindrome number? {check_palindrome_number(num)}")
```

# 42. Write a program to replace all zeros in a number n with the digit 5.

```python
def replace_zero_with_5(n: int) -> int:
    new_num = 0
    num_of_digits = len(str(n))
    
    while num_of_digits > 0:
        
        # get the first digit
        digit = n // (10 ** (num_of_digits - 1))
        
        # get the remainder ie. number without the first digit
        n %= (10 ** (num_of_digits - 1))

        # replace digit 0 with digit 5
        if digit == 0:
            digit = 5
            
        new_num = new_num * 10 + digit
        
        num_of_digits -= 1
    return new_num

num = 2040608
print(f"Earlier number = {num}")
print(f"After replacing new number = {replace_zero_with_5(num)}")
```

# 46. Write a program to display the first n prime numbers.

```python
def first_n_prime(n: int) -> int:
    if n > 0:
        count =  n
        num = 2
        while count > 0:
            if num == 2 or num == 3:
                print(num, end=" ")
                num += 1
                count -= 1
                continue

            if num % 2 == 0 or num % 3 == 0:
                num += 1
                continue

            divisor = 5
            prime_flag = True
            while divisor * divisor <= num:
                if num % divisor == 0 or num % (divisor + 2) == 0:
                    prime_flag = False
                    break
                divisor += 6
            if prime_flag:
                print(num, end=" ")
                count -= 1
            num += 1      
    else:
        print("Enter a natural number.")

num = int(input("Enter the number of prime number you want: "))
print(f"First {num} prime numbers are :")
first_n_prime(num)
```

```python
def HCF(num1, num2):
    
    factors_num1 = []
    factors_num2 = []
    
    for i in range(1, num1 + 1):
        if num1 % i == 0:
            factors_num1.append(i)
    
    for j in range(1, num2 + 1):
        if num2 % j == 0:
            factors_num2.append(j)

    intersection = list(set(factors_num1) & set(factors_num2))
    return intersection[-1]

print(HCF(24, 36))
print(HCF(3, 5))
```

# 67. Write a program to print Pascal's triangle for n rows.

    1
    1 1
    1 2 1
    1 3 3 1
    1 4 6 4 1
    1 5 10 10 5 1
    1 6 15 20 15 6 1
    ...

Steps:

    1. 1st row of pascal triangle contain 1.
    2. from 2nd row onwards 
    2.1 each row start with 1 and end with 1.
    2.2 Calculate next element till current row 2nd last element: curr_row_next_element[postion] =  prev_row[postion - 1] + prev_row[postion]
    

```python
def generate_pascal_triangle(n):

    # initialize the pascal triangle with the first row having 1 as single element
    pascal_triangle = [[1]]

    for row in range(1, n):
        
        # initialize the curr_row with 1
        curr_row = [1]

        prev_row = pascal_triangle[row-1]
        for position in range(1, len(prev_row)):
            curr_row_next_element = prev_row[position - 1] + prev_row[position]
            curr_row.append(curr_row_next_element)

        # end the curr_row with 1
        curr_row.append(1)
        pascal_triangle.append(curr_row)

    return pascal_triangle

generate_pascal_triangle(7)
```

    iterate through the string

```python
def changecase(text, case):
    output_string = ""
    if case == "UPPER": # for converting text to UPPER CASE
        for char in text:
            if char.isalpha():
                if char.islower():
                    new_char = chr(ord(char) - 32)
                    output_string += new_char
                else:
                    output_string += char
            else:
                output_string += char
    elif case == "lower":
        for char in text:
            if char.isalpha():
                if char.isupper():
                    new_char = chr(ord(char) + 32)
                    output_string += new_char
                else:
                    output_string += char
            else:
                output_string += char
        
    return output_string

s = 'arVinD iS a GooD peRSon! Who is goNNa be ricH.'
case = "UPPER"
print(changecase(s, case))
s2 = 'arVinD iS a GooD peRSon! Who is goNNa be ricH.'
case2 = 'lower'
print(changecase(s2, case2))
```

# 76. Write a program to check whether two strings are anagrams of each other.

```python
def anagramscheck(string1, string2):
    # remove all spaces from both strings
    string1_without_spaces = ''
    for char in string1:
        if char != ' ':
            string1_without_spaces += char

    string2_without_spaces = ''
    for char in string2:
        if char != ' ':
            string2_without_spaces += char

    # convert both strings to lowercase
    string1_without_spaces_lowercase = string1_without_spaces.lower()
    string2_without_spaces_lowercase = string2_without_spaces.lower()

    # Conditions of both the strings to be anagrams of each other are 1. Same lenght 2. Identical characters 3. same character frequencies.

    # check if length of both the strings are same or not.
    if len(string1_without_spaces_lowercase) != len(string2_without_spaces_lowercase):
        return False

    freq_map_string1 = {}
    for char in string1_without_spaces_lowercase:
        freq_map_string1[char] = freq_map_string1.get(char, 0) + 1

    freq_map_string2 = {}
    for char in string2_without_spaces_lowercase:
        freq_map_string2[char] = freq_map_string2.get(char, 0) + 1
        
    # check if identical characters are their in both the strings or not.
    if freq_map_string1.keys() != freq_map_string2.keys():
        return False
    # check if same character frequencies are their or not.
    else:
        for char in freq_map_string1.keys():
            if freq_map_string1.get(char) != freq_map_string2.get(char):
                return False
                
    return True

s1, s2 = 'Listen', 'Silent'
print(anagramscheck(s1, s2))

s3, s4 = "Slot machines", "Cash lost in me"
print(anagramscheck(s3, s4))
s5, s6 = "look", "Lock"
print(anagramscheck(s5, s6))
s7, s8 = "Apple", "pale"
print(anagramscheck(s7, s8))
```

```python
# using str.replace() function to replace white spaces with nothing and sorted() function to get the sorted list of characters of both the strings.
def checkanagrams(text1, text2):
    text1 = text1.replace(" ", "").lower()
    text2 = text2.replace(" ", "").lower()

    if len(text1) != len(text2):
        return False

    return sorted(text1) == sorted(text2)

s1, s2 = 'Listen', 'Silent'
print(checkanagrams(s1, s2))

s3, s4 = "Slot machines", "Cash lost in me"
print(checkanagrams(s3, s4))
s5, s6 = "look", "Lock"
print(checkanagrams(s5, s6))
s7, s8 = "Apple", "pale"
print(checkanagrams(s7, s8))
```

```python
def firstNonRepeatingChar(string):
    freq_map = {}
    for char in string:
        freq_map[char] = freq_map.get(char, 0) + 1

    for char in freq_map:
        if freq_map[char] == 1:
            return char

    # if none of the character appears once
    return None

text = "swiss"
print(firstNonRepeatingChar(text))
```

```python
def togglecase(string):
    output_string = ""
    for char in string:
        if char.isalpha():
            if char.islower():
                output_string += chr(ord(char) - 32)
            else:
                output_string += chr(ord(char) + 32)
        else:
            output_string += char
    return output_string

text = 'HeLLo, WoRlD! My sElF ArvInD AND I hAve sTarTed WinTER ARC ChaLLenGE.'
print(togglecase(text))
```

```python
def thirdLargest(nums): # pushed on Leetcode too
    if not nums or len(nums) < 3:
        return None

    first_largest = second_largest = third_largest = float('-inf')

    for num in nums:
        # Case 1: current num is the largest found so far
        if num > first_largest:
            third_largest = second_largest
            second_largest = first_largest
            first_largest = num
        # Case 2: current num is in between first and second largest
        elif num > second_largest and num != first_largest:
            third_largest = second_largest
            second_largest = num
        # Case 3: current num is greater than third largest and smaller than both first and second largest
        elif num > third_largest and num != second_largest and num != third_largest:
            third_largest = num

    return third_largest if third_largest != float('-inf') else None
```

```python
def topKFrequent(nums: list[int], k: int) -> list[int]:
    freq_map = dict()
    for num in nums:
        freq_map[num] = freq_map.get(num, 0) + 1

    sorted_freq_map = dict(sorted(freq_map.items(), key=lambda x: x[1], reverse=True))
    return list(sorted_freq_map.keys())[:k]

nums1 = [1,1,1,2,2,3]
k1 = 2
print(topKFrequent(nums1, k1))
nums2 = [1]
k2 = 1
print(topKFrequent(nums2, k2))
nums3 = [1,2,1,2,1,2,3,1,3,2]
k3 = 2
print(topKFrequent(nums3, k3))
```

# 91. Write a program to sort an array in ascending order (bubble sort).

=> Just like bubbles rise in water, similarly the smallest or largest element rise to top of the array depending on ascending order or descending order sorting.

arr = [5,1,4,2], sort arr in ascending order using bubble sort algorithm

Iteration 1:

    1. Compare (5,1), since 5 > 1, swap both => [1,5,4,2]
    2. Compare (5,4), since 5 > 4, swap both => [1,4,5,2]
    3. Compare (5,2, since 5 > 2, swap both => [1,4,2,5]
    Now 5 is at its correct location

Iteration 2:

    1. Compare (1,4), since 1 < 4, no swap => [1,4,2,5]
    2. Compare (4,2), since 4 > 2, swap both => [1,2,4,5]
    Now 4 is at its correct location

Iteration 3:

    1. Compare (1,2), since 1 < 2, no swap => [1,2,4,5]
    Since no swapping occurred the resultant array is sorted ascendingly 

```python
def bubbleSort(arr):

    if not arr: 
        return -1
    if len(arr) == 1:
        return arr # since a one-element array is always sorted.

    for i in range(len(arr)):
        swapped = False

        for j in range(len(arr) - i - 1):
            if arr[j] > arr[j+1]:
                # swap both 
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True

        if not swapped:
            break
    return arr

nums1 = [1,3,4,2,6,9,5]
print(bubbleSort(nums1))
nums2 = [2,3,4,5,6,7,8,9]
print(bubbleSort(nums2))
```

2. Iterate through the total of (length(arr1) + length(arr2))
3. Take the elements from arr1 one by one and append those elements to output_arr.

# 93. Write a program to find the sum of all even-indexed and odd-indexed elements separately.

# 97. Write a program to count the number of prime digits present in a number n.


```python
def countPrimeDigits(number):
    prime_digits_lookup = [2,3,5,7]
    prime_digit_count = 0
    prime_digits_in_number = []
    while number > 0:
        digit = number % 10
        if digit in prime_digits_lookup:
            prime_digit_count += 1
            prime_digits_in_number.append(digit)
        number //= 10
    return prime_digit_count, prime_digits_in_number

n1 = 123456789
print(countPrimeDigits(n1))
```

```python
def checkPalindrome(num):
    temp = num
    reversed_num = 0
    while temp > 0 :
        digit = temp % 10
        reversed_num = reversed_num * 10 + digit
        temp //= 10
    return reversed_num == num

def checkPrime(num):
    if num <= 1:
        return False
    if 1 < num < 4:
        return True
    if num % 2 == 0 or num % 3 == 0:
        return False
    count = 5
    while count * count <= num:
        if (num % count == 0) or ( num % (count + 2) == 0):
            return False
        count += 6
    return True

def checkBothPrimeAndPalindrome(num):
    return checkPrime(num) and checkPalindrome(num)

n1 = 111
print(checkPalindrome(n1))
print(checkPrime(n1))
print(checkBothPrimeAndPalindrome(n1))
n2 = 919
print(checkPalindrome(n2))
print(checkPrime(n2))
print(checkBothPrimeAndPalindrome(n2))
n3 = 29
print(checkPalindrome(n3))
print(checkPrime(n3))
print(checkBothPrimeAndPalindrome(n3))
```

# 103. Minimum steps to convert an Array into permutation of numbers from 1 to N (Using Addition/Subtraction)

```python
def arr_permutation(arr):
    n = len(arr)

    # Sort the given array
    sorted_arr = sorted(arr)

    # minimum operations
    min_ops = 0

    for num in range(n):
        min_ops += abs(sorted_arr[num] - (num + 1))

    return min_ops

nums = [0, 2, 3, 4, 1, 6, 8, 9]
print(arr_permutation(nums))
```

# 104. Change the array into a permutation of numbers from 1 to n

```python
def changeArrToPermutationByReplacing(arr):
    n = len(arr)
    freq_map = {}

    for num in arr:
        freq_map[num] = freq_map.get(num, 0) + 1

    min_ops = 0
    for num, count in freq_map.items():

        if num < 1 or num > n:
            min_ops += count
        else:
            min_ops += (count - 1)

    return min_ops

nums = [2,2,3,7] # need to replace 2 -> 1 and 7 -> 4 => output= 2
print(changeArrToPermutationByReplacing(nums))
nums1 = [0, 2, 3, 4, 1, 6, 8, 9] # need to replace 0 -> 5 and 9 -> 7 => output = 2
print(changeArrToPermutationByReplacing(nums1))
```
