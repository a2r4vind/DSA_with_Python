```python
# int(logarithm base 10 (num) + 1) = digit_counts  -> TC = O(1), SC = O(1)
import math
def countDigits(num):
    return int(math.log10(num) + 1)

print(countDigits(5873))
print(countDigits(1122334455))
```

```python
# Check Palindrome
n = 1234
def checkPalindrome(n): # TC = O(log10(N)) and SC = O(1)
    num = n
    rev_num = 0
    while num > 0:
        last_digit = num % 10
        rev_num = rev_num * 10 + last_digit
        num //= 10
    # return True if rev_num == n else False
    return rev_num == n

checkPalindrome(n)
```

```python
# Check Armstrong number 
def checkArmstrong(n): # TC = O(log10(N)) and SC = O(1)
    # num = n
    # digit_counts = 0
    # while num > 0:
    #     digit_counts += 1
    #     num //= 10
    digit_counts = len(str(n))
    num = n
    result = 0
    while num > 0:
        last_digit = num % 10
        result = result + (last_digit ** digit_counts)
        num //= 10

    return result == n

print(checkArmstrong(153))
print(checkArmstrong(1634))
```

```python
# print factors/divisors of a number 

# Brute force
def factorsOfNumber(n): # TC = O(N), SC = O(k) , where k = total number of factors
    factors = []
    for divisor in range(1, n+1):
        if n % divisor == 0:
            factors.append(divisor)
    return factors

factorsOfNumber(100)
```

```python
# a bit optimized than brute force
# iterate till the half of the given number ie. n//2

def findfactors(n): # TC = O(N/2) approx O(N) because iterate till n/2
    result = []
    for div in range(1, n//2 + 1):
        if n % div == 0:
            result.append(div)
    result.append(n) # appending the number itself
    return result

findfactors(10)
```

# Optimal solution to find the factors of a number.

iterate till sqrt(given_num)

    num = 36

    2 --- 18 => 36 / 2 = 18
    3 --- 12 => 36 / 3 = 12
    4 --- 9 => 36 / 4 = 9
    6 --- 6 => 36 / 6 = 6
    

```python
from math import sqrt
def findallfactors(n):
    result = [] # SC = O(k) , where k is total number of factors
    for divisor in range(1, int(sqrt(n)) + 1): # TC = O(sqrt(N))
        if n % divisor == 0:
            result.append(divisor)
            if n // divisor != divisor:
                result.append(n//divisor)
    # return result.sort() # this will sort the results and this will add the TC = O(N.log(N))

    return result # returned results factor lists will not be in ordered

print(findallfactors(36))
```

# Frequency map / Dictionary 

```python
# method 1
nums = [5,6,7,7,1,9,111, 1, 1, 5, 1, 1]
freq_map = dict()
for idx in range(0, len(nums)): # TC = O(N), SC = O(N)
    if nums[idx] in freq_map: # TC = O(1)
        freq_map[nums[idx]] += 1 # TC = O(1)
    else:
        freq_map[nums[idx]] = 1 # TC = O(1)


freq_map
```

```python
# method 2
nums = [5,6,7,7,1,9,111, 1, 1, 5, 1, 1]

hash_map = {}
n = len(nums)
for i in range(0,n):
    hash_map[nums[i]] = hash_map.get(nums[i],0) + 1

hash_map
    
```

# Hashing 

```python
n = [5,3,2,2,1,5,5,7,5,10]
m = [10,111,1,9,5,65,2]
# constraints
# 1 <= n[i] <= 10
```

```python
hash_list = [0] * 11 # initialize the hash list

for num in n:
    hash_list[num] += 1

for num in m:
    if num < 1 or num > 10:
        print(f"{num}: {0}")
    else:
        print(f"{num}: {hash_list[num]}")
```

```python
# dictionary_method/freq_map
freq_map = dict()

for i in range(len(n)):
    freq_map[n[i]] = freq_map.get(n[i],0) + 1

for num in m:
    if num < 1 or num > 10:
        print(f"{num}: {0}")
    else:
        print(f"{num}: {freq_map.get(num,0)}")
```

```python
s = "azyxyyzaaaa"
q = ["d", "a", "y", "x"]
# constraints
# 'a' <= s[i] <= 'z' ie. 97 - 122
```

```python
hash_list = [0] * 26

for char in s:
    ascii_value = ord(char)
    idx = ascii_value - 97
    hash_list[idx] += 1

for char in q:
    ascii_value = ord(char)
    idx = ascii_value - 97
    print(f"{char}: {hash_list[idx]}") 
```
