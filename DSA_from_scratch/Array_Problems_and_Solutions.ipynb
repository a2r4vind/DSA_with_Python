# Q. Largest element in an Array.

```python
def largest_element(arr):
    n = len(arr)
    largest = float('-inf')
    for i in range(n):
        largest = max(largest, arr[i])

    return largest

nums = [55, 32, -97, 99, 3, 67]
print(largest_element(nums))
```

# Q. Find 2nd largest element (without sorting)

```python
# Brute force solution includes sorting.
def Sec_largest(arr):
    arr.sort() # Time complexity => O(NlogN) | Space complexity => O(1) 
    return arr[-2]
```

```python
print(Sec_largest([88, 99, 98]))
```

```python
# Better solution 
def Second_largest(nums):
    largest = sec_largest = float('-inf')
    n = len(nums)
    for i in range(n):
        largest = max(largest, nums[i])

    for i in range(n):
        if nums[i] > sec_largest and nums[i] != largest:
            sec_largest = nums[i]
    return sec_largest
```

```python
print(Second_largest([88, 99, 98]))
```

```python
# Optimal solution
def second_largest(nums):
    n = len(nums)
    largest = sec_largest = float('-inf')
    for i in range(n):
        if nums[i] > largest:
            sec_largest = largest
            largest = nums[i]
        elif nums[i] > sec_largest and nums[i] != largest:
            sec_largest = nums[i]

    return sec_largest
```

```python
print(second_largest([45, 56, 27, 94, 89, 111]))
```

# Q. Check if Array is sorted?

```python
def is_sorted(nums): # only checks if given array is sorted in ascending order or not
    n = len(nums)
    for i in range(0, n-1):
        if nums[i] > nums[i+1]:
            return False
    return True
        
```

```python
is_sorted([1,2,3,4,5,6,7,8])
```

```python
# if the order in which given array is sorted is not given and we need to check whether the given array is sorted or not?
def check_sorted(nums):
    n = len(nums)
    for i in range(1, n-1):
        if nums[i-1] < nums[i] > nums[i+1]:
            return False
        elif nums[i] > nums[i] < nums[i+1]:
            return False
    return True
```

```python
check_sorted([1,2,3,4,5,6,7,8])
```

```python
check_sorted([1,2,5,8,3,10,14,15])
```

```python
check_sorted([15,14,10,3,8,5,2,1])
```
