# Q. Largest element in an Array.

```python
def largest_element(arr):
    n = len(arr)
    largest = float('-inf')
    for i in range(n):
        largest = max(largest, arr[i])

    return largest

nums = [55, 32, -97, 99, 3, 67]
print(largest_element(nums))
```

# Q. Find 2nd largest element (without sorting)

```python
# Brute force solution includes sorting.
def Sec_largest(arr):
    arr.sort() # Time complexity => O(NlogN) | Space complexity => O(1) 
    return arr[-2]
```

```python
print(Sec_largest([88, 99, 98]))
```

```python
# Better solution 
def Second_largest(nums):
    largest = sec_largest = float('-inf')
    n = len(nums)
    for i in range(n):
        largest = max(largest, nums[i])

    for i in range(n):
        if nums[i] > sec_largest and nums[i] != largest:
            sec_largest = nums[i]
    return sec_largest
```

```python
print(Second_largest([88, 99, 98]))
```

```python
# Optimal solution
def second_largest(nums):
    n = len(nums)
    largest = sec_largest = float('-inf')
    for i in range(n):
        if nums[i] > largest:
            sec_largest = largest
            largest = nums[i]
        elif nums[i] > sec_largest and nums[i] != largest:
            sec_largest = nums[i]

    return sec_largest
```

```python
print(second_largest([45, 56, 27, 94, 89, 111]))
```

# Q. Check if Array is sorted?

```python
def is_sorted(nums): # only checks if given array is sorted in ascending order or not
    n = len(nums)
    for i in range(0, n-1):
        if nums[i] > nums[i+1]:
            return False
    return True
        
```

```python
is_sorted([1,2,3,4,5,6,7,8])
```

```python
# if the order in which given array is sorted is not given and we need to check whether the given array is sorted or not?
def check_sorted(nums):
    n = len(nums)
    for i in range(1, n-1):
        if nums[i-1] < nums[i] > nums[i+1]:
            return False
        elif nums[i] > nums[i] < nums[i+1]:
            return False
    return True
```

```python
check_sorted([1,2,3,4,5,6,7,8])
```

```python
check_sorted([1,2,5,8,3,10,14,15])
```

```python
check_sorted([15,14,10,3,8,5,2,1])
```

# Q. Remove duplicates in sorted array.

```python
def remove_duplicates(nums):
    n = len(nums)
    freq_map = {}
    for i in range(n): # TC => O(N)
        freq_map[nums[i]] = 0

    j = 0
    for key in freq_map: # TC => O(N) in worst case in which all elements are unique
        nums[j] = key
        j += 1

    return j
```

```python
n1 = [1,1,1,2,3,4,4,7,9,9,9,10]
remove_duplicates(n1)
```

```python
# Optimal Solution --> 2 pointer approach
def remove_Duplicates(nums):
    n = len(nums)
    if n == 1:
        return n
    i = 0
    j = i + 1
    while j < n:
        if nums[j] != nums[i]:
            i += 1
            nums[i], nums[j] = nums[j], nums[i]
        j += 1
    return i + 1
```

```python
n1 = [1,1,1,2,3,4,4,7,9,9,9,10]
remove_Duplicates(n1)
```

# Q. Right-rotate the array by 1 place.


```python
# By slicing
def right_rotate_by_slicing(nums): # TC => O(N) | SC => O(1)
    n = len(nums)
    nums[:] = [nums[-1]] + nums[:n-1]
```

```python
n1 = [5,-2, 3, 9, 0, 6, 10, 7]
right_rotate_by_slicing(n1)
n1
```

```python
# By shifting each element by one on right side
def right_rotate_by_shifting(nums): # TC => O(N) | SC => O(1)
    n = len(nums)
    if n == 1:
        return n
    temp = nums[-1]
    for i in range(n-2, -1, -1):
        nums[i+1] = nums[i]

    nums[0] = temp
```

```python
n1 = [5,-2, 3, 9, 0, 6, 10, 7]
right_rotate_by_shifting(n1)
n1
```

# Q. Right-rotate the array by k space

```python
# using slicing
def right_rotate_by_K(nums,k):
    n = len(nums)
    k = k % n
    nums[:] = nums[n-k:] + nums[:n-k]
```

```python
n1 = [3,9,5,6,7,2,10,9]
right_rotate_by_K(n1,5)
n1
```

```python
# using pop() and insert()
def right_rotate_by_K(nums, k):
    n = len(nums)
    k = k % n
    for _ in range(k):
        e = nums.pop()
        nums.insert(0,e)
```

```python
n1 = [3,9,5,6,7,2,10,9]
right_rotate_by_K(n1,3)
n1
```

```python
# Optimal solution is by using recursion concept
def reverse(nums, left, right):
    while left < right:
        nums[left], nums[right] = nums[right], nums[left]
        left += 1
        right -= 1

def right_rotate(nums, k):
    n = len(nums)
    k = k % n
    reverse(nums, n-k, n-1)
    reverse(nums, 0, n-k-1)
    reverse(nums, 0, n-1)
```

```python
m = [3,9,5,6,7,2,10,9]
right_rotate(m, 5)
m
```

# Q. Move zeros to the end of the array while maintaining the order of non-zero elements and performing in-place changes.

```python
# Brute force solution
def move_zeros_brute_force(nums):
    n = len(nums)
    temp = []
    for i in range(n):
        if nums[i] != 0:
            temp.append(nums[i])
    m = len(temp)
    for i in range(m):
        nums[i] = temp[i]
    for i in range(m, n):
        nums[i] = 0
```

```python
n1 = [1,0,2,4,3,0,0,0,3,5,1]
move_zeros_brute_force(n1)
n1
```

```python
# Optimal solution
def move_zeros_to_end(nums):
    n = len(nums)
    # if single element array
    if n == 1:
        return 
    i = 0
    while i < n:
        if nums[i] == 0:
            break
        i += 1
    # if no zero is present
    if i == n:
        return
        
    j = i + 1
    while j < n:
        if nums[j] != 0:
            nums[i], nums[j] = nums[j], nums[i]
            i += 1
        j += 1
```

```python
n1 = [1,0,2,4,3,0,0,0,3,5,1]
move_zeros_to_end(n1)
n1
```

# Merge two sorted array without duplication

```python
def merge_two_sorted_arr(nums1, nums2):
    result = []
    n, m = len(nums1), len(nums2)
    i, j = 0, 0
    while i < n and j < m:
        if nums1[i] <= nums2[j]:
            if len(result) == 0 or result[-1] != nums1[i]: 
                result.append(nums1[i])
            i += 1
        else:
            if len(result) == 0 or result[-1] != nums2[j]:
                result.append(nums2[j])
            j += 1

    while i < n:
        if len(result) == 0 or result[-1] != nums1[i]:
            result.append(nums1[i])
        i += 1
    while j < m:
        if len(result) == 0 or result[-1] != nums2[j]:
            result.append(nums2[j])
        j += 1

    return result
            
            
```

```python
n1 = [1,1,1,2,4,6,7]
n2 = [1,2,3,6,7,8,9,10]
print(merge_two_sorted_arr(n1,n2))
```
