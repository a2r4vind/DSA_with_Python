# Binary Search => Time Complexity = O(logN)

Condition:

    1. Given sorted array.

```python
# Iterative approach
def binarySearch(nums, target):
    n = len(nums)
    low, high = 0, n - 1
    while low <= high:
        mid = (low + high) // 2
        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

n = [12, 14, 15, 17 , 28, 31, 83, 99]
print(binarySearch(n, 28))
print(binarySearch(n, 100))
```

```python
# recursive approach
def binary_Search(nums, low, high, target):
    if low > high:
        return -1
    mid = (low + high) // 2
    if nums[mid] == target:
        return mid
    elif nums[mid] < target:
        return binary_Search(nums, mid + 1, high, target)
    else:
        return binary_Search(nums, low, mid - 1, target)

n = [12, 14, 15, 17 , 28, 31, 83, 99]
print(binary_Search(n, 0, len(n) - 1, 28))
print(binary_Search(n, 0, len(n) - 1, 100))
```

```python
def floorAndCeil(nums, target):
    n = len(nums)
    low, high = 0, n - 1
    floor, ceil = -1, -1
    while low <= high:
        mid = (low + high) // 2
        if nums[mid] == target:
            return [nums[mid], nums[mid]]
        elif nums[mid] > target:
            ceil = nums[mid]
            high = mid - 1
        else:
            floor = nums[mid]
            low = mid + 1
    return [floor, ceil]

nums = [3, 4, 4, 4, 8, 9, 9, 10, 12, 12, 14, 15]
target = 11
print(floorAndCeil(nums, target))
```

# Q. first and last occurrence in a sorted array.

```python
# brute force solution
def firstAndLast_bruteforce(nums, target):
    n = len(nums)
    if not nums:
        return []
    first, last = -1, -1
    for i in range(n): # Time Complexity = O(N)
        if nums[i] == target:
            if first == -1:
                first = i
            last = i
    return [first, last]

nums = [1, 2, 3, 3, 3, 3, 3 ,5 , 6, 8, 9, 9, 10]
print(firstAndLast_bruteforce(nums, target=3))
print(firstAndLast_bruteforce(nums, target=7))
```

```python
# optimal solution
def lower_bound(nums, target):
    n = len(nums)
    low, high = 0, n - 1
    lowerBound = -1 
    while low <= high:
        mid = (low + high) // 2
        if nums[mid] == target:
            lowerBound = mid
            high = mid - 1
        else:
            low = mid + 1
    return lowerBound
    
def upper_bound(nums, target):
    n = len(nums)
    upperBound = -1
    low, high = 0, n - 1
    while low <= high:
        mid = (low + high) // 2
        if nums[mid] > target:
            upperBound = mid
            high = mid - 1
        else:
            low = mid + 1
    return upperBound
    
def firstAndLast(nums, target):
    lowerBound = lower_bound(nums, target)
    if lowerBound == -1:
        return [-1, -1]
    upperBound = upper_bound(nums, target)
    return [lowerBound, upperBound - 1]

nums = [1, 2, 3, 3, 3, 3, 3 ,5 , 6, 8, 9, 9, 10]
print(firstAndLast(nums, 3))
print(firstAndLast(nums, 7))
```
