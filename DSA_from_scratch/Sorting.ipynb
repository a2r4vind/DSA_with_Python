# 1. Selection sort

Time complexity = O(n(n+1)/2) => O(n^2)

Space complexity = O(1)

```python
def selection_sort(nums):
    n = len(nums)
    for i in range(n):
        min_idx = i
        for j in range(i+1,n):
            if nums[j] < nums[min_idx]:
                min_idx = j
        nums[i], nums[min_idx] = nums[min_idx], nums[i]
        
    return nums

n1 = [8,6,1,7,9,5,2,3]
print(selection_sort(n1))
```

```python
def selection_sort_descending(nums):
    n = len(nums)
    for i in range(n):
        max_idx = i
        for j in range(i+1, n):
            if nums[j] > nums[max_idx]:
                max_idx = j
        nums[i], nums[max_idx] = nums[max_idx], nums[i]
    return nums

n1 = [8,6,1,7,9,5,2,3]
print(selection_sort_descending(n1))
```

# 2. Bubble Sort

Time complexity = O(n(n+1)/2) => O(n^2) | BestCase Time Complexity = O(n)

Space complexity = O(1)

```python
def bubble_sort(nums):
    n = len(nums)
    for i in range(n-2,-1,-1):
        is_swap = False
        for j in range(0, i + 1):
            if nums[j] > nums[j+1]:
                nums[j], nums[j+1] = nums[j+1], nums[j]
    return nums

n1 = [8,6,1,7,9,5,2,3]
print(bubble_sort(n1))
```

# 3. Insertion sort

Time Complexity = O(n(n+1)/2) => O(n^2)

Space Complexity = O(1)

```python
def insertion_sort(nums):
    n = len(nums)
    for i in range(1,n):
        key = nums[i]
        j = i - 1
        while j >= 0 and nums[j] > key:
            nums[j+1] = nums[j]
            j -= 1
        nums[j+1] = key
    return nums

n1 = [8,6,1,7,9,5,2,3]
print(insertion_sort(n1))
```
