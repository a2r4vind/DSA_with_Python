# Selection sort

```python
def selection_sort(nums):
    n = len(nums)
    for i in range(n):
        min_idx = i
        for j in range(i+1,n):
            if nums[j] < nums[min_idx]:
                min_idx = j
        nums[i], nums[min_idx] = nums[min_idx], nums[i]
        
    return nums

n1 = [8,6,1,7,9,5,2,3]
print(selection_sort(n1))
```
