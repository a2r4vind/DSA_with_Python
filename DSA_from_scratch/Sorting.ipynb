# 1. Selection sort

In Selection sort, for ascending order, we **select** the minimum element by its index and swap it into its correct position from the left.

Time complexity = O(n(n+1)/2) => O(n^2)

Space complexity = O(1)

```python
def selection_sort(nums):
    n = len(nums)
    for i in range(n):
        min_idx = i
        for j in range(i+1,n):
            if nums[j] < nums[min_idx]:
                min_idx = j
        nums[i], nums[min_idx] = nums[min_idx], nums[i]
        
    return nums

n1 = [8,6,1,7,9,5,2,3]
print(selection_sort(n1))
```

```python
def selection_sort_1(nums):
    n = len(nums)
    for i in range(0,n):
        min_idx = i
        for j in range(i+1, n):
            if nums[j] < nums[min_idx]:
                min_idx = j

        nums[i], nums[min_idx] = nums[min_idx], nums[i]

    return nums
```

```python
def selection_sort_descending(nums):
    n = len(nums)
    for i in range(n):
        max_idx = i
        for j in range(i+1, n):
            if nums[j] > nums[max_idx]:
                max_idx = j
        nums[i], nums[max_idx] = nums[max_idx], nums[i]
    return nums

n1 = [8,6,1,7,9,5,2,3]
print(selection_sort_descending(n1))
```

# 2. Bubble Sort

In Bubble Sort (ascending order), we compare adjacent elements from left to right and swap them if the left element is greater than the right element; we continue this process until we reach the end of the given nums array. This results in the maximum value element rising to the right of the nums array, just like bubbles rise to the top. 

Time complexity = O(n(n+1)/2) => O(n^2) | BestCase Time Complexity = O(n)

Space complexity = O(1)

```python
def bubble_sort(nums):
    n = len(nums)
    for i in range(n-2,-1,-1):
        is_swap = False
        for j in range(0, i + 1):
            if nums[j] > nums[j+1]:
                nums[j], nums[j+1] = nums[j+1], nums[j]
                is_swap = True
        if not is_swap:
            return nums
            
    return nums

n1 = [8,6,1,7,9,5,2,3]
print(bubble_sort(n1))
```

```python
def bubble_sort_1(nums):
    n = len(nums)
    for i in range(n-2,-1, -1):
        is_swap = False
        for j in range(0,i+1):
            if nums[j] > nums[j+1]:
                nums[j],nums[j+1] = nums[j+1],nums[j]
                is_swap = True
        if not is_swap:
            return nums

    return nums
```

# 3. Insertion sort

In Insertion Sort, we compare the current element with the element(s) present before it and try to find out the correct position of the current element; if found, then we **insert** the current element at its correct index position and shift the remaining elements +1 position ahead. 

Time Complexity = O(n(n+1)/2) => O(n^2)

Space Complexity = O(1)

```python
def insertion_sort(nums):
    n = len(nums)
    for i in range(1,n):
        key = nums[i]
        j = i - 1
        while j >= 0 and nums[j] > key:
            nums[j+1] = nums[j]
            j -= 1
        nums[j+1] = key
    return nums

n1 = [8,6,1,7,9,5,2,3]
print(insertion_sort(n1))
```

# 4. Merge Sort

Time complexity = O(NlogN)

Space complexity = O(N)

```python
def merge_array(left, right):
    result = []
    n, m = len(left), len(right)
    i, j = 0, 0
    while i < n and j < m:
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    if i < n:
        while i < n:
            result.append(left[i])
            i += 1

    if j < m:
        while j < m:
            result.append(right[j])
            j += 1

    return result

def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left_arr = arr[:mid]
    right_arr = arr[mid:]
    left = merge_sort(left_arr)
    right = merge_sort(right_arr)
    return merge_array(left, right)

n1 = [8,6,1,7,9,5,2,3]
print(merge_sort(n1))
```

# 5. Quick Sort

BestCase and AverageCase Time Complexity = O(NlogN)

WorstCase (all elements are same arr=[4,4,4,4]) Time complexity = O(N^2)

Space complexity = O(1)

```python
def partition(nums, low, high):
    pivot = nums[low]
    i, j = low, high
    while i < j:
        while nums[i] <= pivot and i <= high - 1:
            i += 1
        while nums[j] > pivot and j >= low + 1:
            j -= 1
        if i < j:
            nums[i], nums[j] = nums[j], nums[i]
    nums[low], nums[j] = nums[j], nums[low]
    return j

def quick_sort(nums, low, high):
    if low < high:
        p_idx = partition(nums, low, high)
        quick_sort(nums, low, p_idx - 1)
        quick_sort(nums, p_idx + 1, high)

```
