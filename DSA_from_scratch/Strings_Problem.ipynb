# Q. Remove outermost parenthesis.

```python
def removeOutermostParenthesis(text: str) -> str:
    result = ""
    count = 0
    for ch in text:
        if ch == "(":
            count += 1
            if count > 1:
                result += ch
        else:
            count -= 1
            if count > 0:
                result += ch
    return result

text = "(()())(()(()))(())"
print(removeOutermostParenthesis(text))
```
