# Q. Remove outermost parenthesis.

```python
def removeOutermostParenthesis(text: str) -> str:
    result = ""
    count = 0
    for ch in text:
        if ch == "(":
            count += 1
            if count > 1:
                result += ch
        else:
            count -= 1
            if count > 0:
                result += ch
    return result

text = "(()())(()(()))(())"
print(removeOutermostParenthesis(text))
```

# Q. Reverse words in a given string.

```python
def reverseWords(text):
    words = text.split()
    words.reverse()
    result = " ".join(words)
    return result

string1 = "The sky is blue"
print(reverseWords(string1))
```

# Q. Largest odd number in the string.

```python
def largestOddNum(number: str) -> str:
    n = len(number)

    for i in range(n-1, -1, -1):
        if int(number[i]) % 2 != 0:
            return number[:i+1]
    return ""


nums1 = "52"
nums2 = "86482"
nums3 = "6783862"
nums4 = "546827"
print(largestOddNum(nums1))
print(largestOddNum(nums2))
print(largestOddNum(nums3))
print(largestOddNum(nums4))
```
